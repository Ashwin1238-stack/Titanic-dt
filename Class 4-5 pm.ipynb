## REMOVE OUTLIERS

### IQR METHOD

```python
a=df.select_dtypes(include=['int'])
a
```

```python
for i in a:
    q1=a[i].quantile(0.25)
    q2=a[i].quantile(0.5)
    q3=a[i].quantile(0.75)
    IQR=q3-q1
    Uw=q3+1.5*IQR
    LW=q1-1.5*IQR
    outliers=((a[i]>Uw)|(a[i]<LW))
    print(f"BEFORE OUTLIERS IN {i} - ",outliers.sum())
    a.loc[outliers,i]=q2
    print(f"AFTER OUTLIERS IN {i} - ",((a[i]>Uw)|(a[i]<LW)).sum())
```
